```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
class SingleHeadSelfAttention(nn.Module):
    def __init__(self, k):
        """
        Args:
            k (int): The embedding dimension.
        """
        super().__init__()
        self.k = k

        self.to_queries = nn.Linear(k, k, bias=False)
        self.to_keys = nn.Linear(k, k, bias=False)
        self.to_values = nn.Linear(k, k, bias=False)

    def forward(self, x):
        """
        Args:
            x (torch.Tensor): Input tensor of shape (b, t, k)
                              b: batch size
                              t: sequence length
                              k: embedding dimension
        """
        b, t, k = x.size()

        queries = self.to_queries(x)
        keys = self.to_keys(x)
        values = self.to_values(x)

        dot = torch.bmm(queries, keys.transpose(1, 2))  # Q . K

        dot = dot / (k ** 0.5)                          # scaling

        attention_weights = F.softmax(dot, dim=2)       # (b ,t, t)
        
        output = torch.bmm(attention_weights, values)   # (b, t, t) . (b, t, k) = (b, t, k)

        return output
```
